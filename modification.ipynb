```python
import pandas as pd

# Load the dataset
file_path = "cleaned_german_data.xlsx"
df = pd.read_excel(file_path)

# Display basic information about the dataset
df.info(), df.head()
```

```python
# Rename the columns
column_names = {
    "Col1": "Status_of_existing_checking_account",
    "Col2": "Duration_in_month",
    "Col3": "Credit_history",
    "Col4": "Purpose",
    "Col5": "Credit_amount",
    "Col6": "Savings_account_bonds",
    "Col7": "Present_employment_since",
    "Col8": "Installment_rate",
    "Col9": "Personal_status_and_sex",
    "Col11": "Other_debtors_guarantors",
    "Col12": "Present_residence_since",
    "Col13": "Age_in_years",
    "Col14": "Other_installment_plans",
    "Col15": "Housing",
    "Col16": "Number_of_existing_credits",
    "Col18": "Job",
    "Col20": "Number_of_people_liable",
    "Col21": "Credit_risk"
}

df.rename(columns=column_names, inplace=True)

# Display the first few rows after renaming
df.head()
```

```python
# Define mappings for categorical variables based on German Credit Dataset documentation

checking_account_status_map = {
    "A11": "No checking account",
    "A12": "Less than 200 DM",
    "A13": "More than 200 DM",
    "A14": "No checking account (in another bank)"
}

credit_history_map = {
    "A30": "No credit taken",
    "A31": "All credits paid duly",
    "A32": "Existing credit paid duly",
    "A33": "Delay in past payments",
    "A34": "Critical account (other credits existing)"
}

purpose_map = {
    "A40": "Car (new)",
    "A41": "Car (used)",
    "A42": "Furniture/equipment",
    "A43": "Radio/television",
    "A44": "Domestic appliances",
    "A45": "Repairs",
    "A46": "Education",
    "A47": "Vacation",
    "A48": "Retraining",
    "A49": "Business",
    "A410": "Other"
}

savings_account_map = {
    "A61": "No savings account",
    "A62": "Less than 100 DM",
    "A63": "100 to 500 DM",
    "A64": "500 to 1000 DM",
    "A65": "More than 1000 DM"
}

employment_map = {
    "A71": "Unemployed",
    "A72": "Less than 1 year",
    "A73": "1 to 4 years",
    "A74": "4 to 7 years",
    "A75": "More than 7 years"
}

personal_status_map = {
    "A91": "Male: divorced/separated",
    "A92": "Female: divorced/separated/married",
    "A93": "Male: single",
    "A94": "Male: married/widowed",
    "A95": "Female: single"
}

other_debtors_map = {
    "A101": "None",
    "A102": "Co-applicant",
    "A103": "Guarantor"
}

residence_map = {
    "A121": "Less than 1 year",
    "A122": "1 to 2 years",
    "A123": "2 to 3 years",
    "A124": "More than 3 years"
}

installment_plan_map = {
    "A141": "Bank",
    "A142": "Stores",
    "A143": "None"
}

housing_map = {
    "A151": "Own",
    "A152": "Rent",
    "A153": "For free"
}

job_map = {
    1: "Unskilled",
    2: "Unskilled (resident)",
    3: "Skilled",
    4: "Highly qualified"
}

liable_people_map = {
    "A201": "One",
    "A202": "Two or more"
}

credit_risk_map = {
    1: "Good credit risk",
    2: "Bad credit risk"
}

# Apply mappings to categorical columns
df.replace({
    "Status_of_existing_checking_account": checking_account_status_map,
    "Credit_history": credit_history_map,
    "Purpose": purpose_map,
    "Savings_account_bonds": savings_account_map,
    "Present_employment_since": employment_map,
    "Personal_status_and_sex": personal_status_map,
    "Other_debtors_guarantors": other_debtors_map,
    "Present_residence_since": residence_map,
    "Other_installment_plans": installment_plan_map,
    "Housing": housing_map,
    "Job": job_map,
    "Number_of_people_liable": liable_people_map,
    "Credit_risk": credit_risk_map
}, inplace=True)

# Display the first few rows after replacing categorical values
df.head()
```

```python
df.to_excel("cleaned_german_data_modified.xlsx", index=False)
```
